```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5,6,7]
y=[5,15,10,30,25,20,35]
plt.plot(x,y)
plt.show()
```

```python
x=['mon','tue','wed','thu','fri','sat','sun']
y=[12,45,28,23,42,35,40]
plt.plot(x,y,linestyle='dashed',linewidth=2,color='#120952',label='Bakery sales',marker='o')
plt.title('Bakery sales this week!')
plt.xlabel('Day of the week!')
plt.ylabel('sales per day')
plt.legend(loc='upper left',fontsize=7)
plt.grid(color='#c3f4e7',linestyle=':',linewidth=1)
#plt.xlim('mon','fri')
#plt.ylim(12,35)
plt.show()
```

```python
product=['a','b','c','d']
sales=[1000,900,1200,1100]
plt.bar(product,sales,color='blue',label='sales in 2025')
plt.xlabel('product')
plt.ylabel('sales')
plt.title('sales in 2025')
plt.legend(loc='upper left',fontsize=8)
plt.show()
```

```python
region=['North','East','West','South']
revinue=[1200,2340,1300,2100]
plt.pie(revinue,labels=region,autopct='%1.1f%%',colors=['#1342e7','#99e713','#e713d2','#e71330'])
plt.title("Revinue per region")
plt.show()
```

```python
scores=[37,76,56,98,56,89,45,67,24,76,98,45,34,69,56,46,93,86,78,79]
plt.hist(scores,bins=5,color='#2363ec',edgecolor='black')
plt.title("Score distribution of studends")
plt.xlabel('scores range')
plt.ylabel('student distribution')
plt.show()
```

```python
hours_study=[1,2,3,4,5,6,7]
exam_score=[56,65,76,89,90,91,92]
plt.scatter(hours_study,exam_score,color='#f10b45',marker='o',label="Student data")
plt.xlabel('hours_study')
plt.ylabel('exam_score')
plt.legend()
plt.title("Reletion between study time and exam score")
plt.grid()
plt.show()
```

```python
plt.scatter([1,2,3,4],[56,76,84,90],color='blue',marker='o',label="Class A")
plt.scatter([1,2,3,4],[58,75,87,98],color='red',marker='o',label="Class B")
plt.xlabel('hours_study')
plt.ylabel('exam_score')
plt.legend()
plt.title("Comparison of two classes")
plt.grid(True)
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,20,15,5,20]
plt.subplot(1,2,1)
plt.plot(x,y)
plt.title('line chart')
plt.subplot(1,2,2)
plt.bar(x,y)
plt.title('bar chart')
plt.tight_layout()
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,20,15,5,20]
fig,ax=plt.subplots(2,2,figsize=(15,10))                   
ax[0,0].plot(x,y,color='orange')
ax[0,0].set_title('Line chart')
ax[0,1].bar(x,y,color='red')
ax[0,1].set_title('Bar chart')
ax[1,0].scatter(x,y,color='green')
ax[1,0].set_title('Scatter chart')
ax[1,1].pie(y,labels=x,colors=['#2d06a5','#63dd05','#a506b0','#328cf5','#f4dc35'])
ax[1,1].set_title('Pie chart')
fig.suptitle("Comparison of charts")
plt.tight_layout()
plt.savefig('matplotlib_charts.png',dpi=300,bbox_inches='tight')
plt.show()
```
